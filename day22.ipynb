```python
import numpy as np
import pandas as pd
```

```python
url=("https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv")
df=pd.read_csv(url)
df
```

```python
#Display: shape, head, column names, datatypes, missing value
print(df.shape)
print(df.head())
print(df.columns)
print(df.isnull())
```

```python
df.isna()
```

```python
print(df.dtypes)
```

```python
#Check and remove duplicates
print(df.duplicated)
```

```python
#Rename columns ("Title" > "Movie_Title"
```

```python
df.rename(columns={"Title":"Movie_Title"},inplace=True)
```

```python
df.columns
```

```python
#Get unique genres
df["Genre"].nunique()
```

```python
#Extract number of genres per movie
df["Genre_count"]=df["Genre"].str.split(",").str.len()
df[["Movie_Title","Genre","Genre_count"]].head()
```

```python
df["Revenue (Millions)"].isnull().sum()
```

```python
df["Metascore"].isnull().sum()
```

```python
#Handle missing values in 'Revenue (Millions)' and 'Metascore
df["Revenue (Millions)"].fillna(df["Revenue (Millions)"].mean(),inplace=True)
df["Revenue (Millions)"].isnull().sum()
```

```python
df["Metascore"].fillna(df["Metascore"].mean(),inplace=True)
df["Metascore"].isnull().sum()
```

```python
#Top 5 movies by Rating
top_5=df.sort_values(by="Rating",ascending=False).head()
print(top_5[["Movie_Title","Rating"]])

```

```python
#Movies with highest/lowest revenue
high_revenue=df.sort_values(by="Revenue (Millions)",ascending=False).head(1)
print(high_revenue[["Movie_Title","Revenue (Millions)"]])

```

```python
low_revenue=df.sort_values(by="Revenue (Millions)",ascending=True).head(1)
print(low_revenue[["Movie_Title","Revenue (Millions)"]])
```

```python
#Avg. rating by Director
df.groupby("Director")["Rating"].mean().sort_values(ascending=False)
```

```python
#Movies per year
df.groupby("Year")["Movie_Title"].count()
```

```python
#Movies with rating > 8 & revenue > 100M
df[(df["Rating"]>8)&(df["Revenue (Millions)"]>100)]
```

```python
#Group by Director: avg Rating, Revenue
df.groupby("Director")[["Rating","Revenue (Millions)"]].mean()
```

```python
#Group by Year: count, avg Rating
df.groupby("Year").agg({
    "Movie_Title":"count",
    "Rating":"mean"
})
```

```python
#Create a new column Profitability:
#a) Revenue divided by Duration
#b) Find top 10 most profitable movies
df["Profitability"]=df["Revenue (Millions)"]/df["Runtime (Minutes)"]
df
```

```python
df.sort_values(by="Profitability",ascending=False).head()
```

```python
#Create a column that categorizes movies by rating:
#a) Rating < 6: Poor
#b) 6 ≤ Rating < 7.5: Average
#c) Rating ≥ 7.5: Excellent
df["Rating_Category"]=np.where(
    df["Rating"]<6,"poor",
    np.where(df["Rating"]<7.5,"Average","Excellent")
)
df
```

```python
#Create a custom function to score movies based on:
#i) Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score
#Apply a function to dataset to assign the score
def calculate_score(row):
  return row["Rating"]*10 + row["Revenue (Millions)"]
df["High_score"]=df.apply(calculate_score,axis=1)
df
```
