```python
import numpy as np
import pandas as pd
np.random.seed(42)
n = 200
df = pd.DataFrame({
"order_id": range(1001, 1001 + n),
"order_date": pd.date_range("2024-01-01", periods=n, freq="D"),
"customer_name": np.random.choice(
["Anita", "Ravi", "Meera", "Arjun", "Sneha", "Karthik", "Divya",
"Suresh"], n),
"city": np.random.choice(
["Kochi", "Chennai", "Bengaluru", "Hyderabad", "Mumbai"], n),
"category": np.random.choice(
["Electronics", "Clothing", "Grocery", "Books", "Furniture"], n),
"quantity": np.random.randint(1, 10, n),
"unit_price": np.random.randint(100, 5000, n),
"payment_mode": np.random.choice(
["UPI", "Card", "Cash", "NetBanking"], n),
})
df.to_csv("sales_data.csv", index=False)
print("File created with", len(df), "rows")
```

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect("sales.db")
cursor = conn.cursor()
cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    order_date TEXT,
    customer_name TEXT,
    quantity INTEGER,
    city TEXT,
    category TEXT,
    unit_price INTEGER,
    payment_mode TEXT
)
""")

print("Database and table created successfully!")
```

```python
#E3 1
query1 = """
SELECT *
FROM orders
WHERE city = 'Mumbai'
"""
mumbai_orders = pd.read_sql_query(query1, conn)
print(mumbai_orders)
```

```python
#E3 2
query2 = """
SELECT category, SUM(quantity) AS total_quantity
FROM orders
GROUP BY category
"""
category_quantity = pd.read_sql_query(query2, conn)
print(category_quantity)
```

```python
#e3 3
query3 = """
SELECT city, AVG(unit_price) AS average_price
FROM orders
GROUP BY city
HAVING AVG(unit_price) > 2500
"""

high_price_cities = pd.read_sql_query(query3, conn)

print(high_price_cities)
```

```python
#E3 4
query4 = """
SELECT *,
       quantity * unit_price AS total_amount
FROM orders
ORDER BY total_amount DESC
LIMIT 5
"""
top_orders = pd.read_sql_query(query4, conn)
print(top_orders)
```

```python
import pandas as pd

def extract(file_path):
    df = pd.read_csv(file_path)

    print("Number of rows loaded:", len(df))

    return df
```

```python
sales = extract("sales_data.csv")
```

```python
#F2
def transform(df):
    df["total_amount"] = df["quantity"] * df["unit_price"]
    def get_order_size(amount):
        if amount >= 15000:
            return "Large"
        elif amount >= 5000:
            return "Medium"
        else:
            return "Small"

    df["order_size"] = df["total_amount"].apply(get_order_size)
    df = df.drop_duplicates(subset="order_id")
    df = df[[
        "order_id",
        "order_date",
        "city",
        "category",
        "total_amount",
        "order_size"
    ]]

    return df
```

```python
cleaned_sales = transform(sales)
print(cleaned_sales)
```

```python
print(cleaned_sales.shape)
```

```python
print(cleaned_sales.head())
```

```python
#f3
import sqlite3

def load(df, db_name, csv_name):

    conn = sqlite3.connect(db_name)

    df.to_sql(
        "clean_orders",
        conn,
        if_exists="replace",
        index=False
    )

    conn.close()

    df.to_csv(csv_name, index=False)

    print("Data successfully saved to database and CSV file!")
```

```python
load(cleaned_sales, "sales.db", "clean_orders.csv")
```

ETL stands for Extract, Transform and Load. In this process, data is extracted from the source, transformed or cleaned, and then loaded into the destination

ELT stands for Extract, Load and Transform. In this process, raw data is first loaded into the destination and transformed after that

here in this assessment,it is ETL because I extracted data from a CSV file, transformed it using Python and Pandas, and then loaded the cleaned data into a SQLite database and a new CSV file
